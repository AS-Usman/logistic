Task 1: Explore Machine Learning Datasets
Dataset 1
Dataset Name

Titanic Survival Prediction

Dataset Source

https://www.kaggle.com/c/titanic

Problem Statement

Predict whether a passenger survived the Titanic disaster based on information such as age, gender, ticket class, fare, and family members.

Machine Learning Problem Type

Classification

Number of Features (Columns)

12 Features

Target Variable

Survived

Reason for Selecting the Dataset

This is one of the most popular beginner datasets. It is easy to understand and helps learn data preprocessing, feature engineering, and classification algorithms.

Dataset 2
Dataset Name

Heart Disease Dataset

Dataset Source

https://archive.ics.uci.edu/ml/datasets/Heart+Disease

Problem Statement

Predict whether a patient has heart disease based on medical information such as age, blood pressure, cholesterol level, chest pain type, and heart rate.

Machine Learning Problem Type

Classification

Number of Features (Columns)

14 Features

Target Variable

Presence of Heart Disease

Reason for Selecting the Dataset

This dataset is useful because it solves a real healthcare problem and is widely used for learning classification techniques.

Dataset 3
Dataset Name

California Housing Prices

Dataset Source

https://www.kaggle.com/datasets/camnugent/california-housing-prices

Problem Statement

Predict the median house price in different areas of California using housing-related information.

Machine Learning Problem Type

Regression

Number of Features (Columns)

10 Features

Target Variable

Median House Value

Reason for Selecting the Dataset

This dataset is ideal for learning regression algorithms and understanding how different factors affect house prices.

Task 3: Final Dataset Selection
Selected Dataset

Heart Disease Dataset

Justification

I selected the Heart Disease Dataset because it works on a real-world healthcare problem. It contains meaningful medical information and is suitable for learning data preprocessing, feature selection, visualization, and classification algorithms. It also provides good accuracy using machine learning models and helps understand how AI can support healthcare.
