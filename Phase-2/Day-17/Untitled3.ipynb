```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside notebook
%matplotlib inline

# Better plot style
sns.set(style="whitegrid")

print("Libraries Imported Successfully!")
```

```python
# Load Titanic dataset
df = sns.load_dataset("titanic")

# Display first five rows
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names")
print(df.columns)

print("\nDataset Information")
df.info()

print("\nMissing Values")
print(df.isnull().sum())
```

```python
df.describe(include='all')
```

```python
# Average Fare by Passenger Class

average_fare = df.groupby("pclass")["fare"].mean()

plt.figure(figsize=(8,5))

plt.plot(
    average_fare.index,
    average_fare.values,
    marker='o',
    linewidth=3,
    color='blue',
    label='Average Fare'
)

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.legend()
plt.grid(True)

plt.show()
```

## Line Plot Analysis

### Why this chart?

A line plot is useful for showing trends across ordered categories.

### What does it represent?

It shows the average ticket fare for each passenger class.

### Insights

- First Class passengers paid the highest fares.
- Third Class passengers paid the least.
- Ticket fare decreases as passenger class decreases.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="class",
    y="fare",
    hue="sex",
    palette="viridis"
)

plt.title("Average Fare by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.legend(title="Gender")
plt.grid(True)

plt.show()
```

## Bar Chart Analysis

### Why this chart?

Bar charts compare averages across different categories.

### What does it represent?

Average fare paid by males and females in each passenger class.

### Insights

- First Class passengers paid the highest fares.
- Fare differences between males and females are relatively small.
- Third Class passengers had the lowest average fares.

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["age"].dropna(),
    bins=20,
    color="orange",
    edgecolor="black"
)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.grid(True)

plt.show()
```

## Histogram Analysis

### Why this chart?

Histograms display the distribution of numerical data.

### What does it represent?

Distribution of passenger ages.

### Insights

- Most passengers were between 20 and 40 years old.
- Very few passengers were above 70 years.
- The age distribution is slightly right-skewed.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    palette="Set1"
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.grid(True)

plt.show()
```

## Scatter Plot Analysis

### Why this chart?

Scatter plots help identify relationships between two numerical variables.

### What does it represent?

Relationship between passenger age and ticket fare.

### Insights

- Higher fares were mostly paid by First Class passengers.
- There is no strong relationship between age and fare.
- Survivors appear across all age groups.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="class",
    y="age",
    palette="pastel"
)

plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")

plt.grid(True)

plt.show()
```

## Box Plot Analysis

### Why this chart?

Box plots identify spread, median, and outliers.

### What does it represent?

Age distribution across passenger classes.

### Insights

- First Class passengers were generally older.
- Third Class had a wider age spread.
- Some age outliers exist in all classes.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="class",
    hue="survived",
    palette="coolwarm"
)

plt.title("Passenger Count by Class and Survival")
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")

plt.legend(title="Survived")

plt.grid(True)

plt.show()
```

## Count Plot Analysis

### Why this chart?

Count plots display the frequency of categorical variables.

### What does it represent?

Number of passengers who survived and did not survive in each passenger class.

### Insights

- First Class had higher survival rates.
- Third Class experienced the highest number of deaths.
- Passenger class strongly influenced survival.

```python
sns.pairplot(
    df,
    vars=["age","fare"],
    hue="survived",
    diag_kind="hist",
    palette="Set2"
)

plt.show()
```

## Pair Plot Analysis

### Why this chart?

Pair plots compare multiple numerical variables simultaneously.

### What does it represent?

Relationships among age, fare, and survival.

### Insights

- Fare shows greater variation than age.
- Survivors tend to have paid higher fares.
- No strong linear relationship exists between age and fare.
